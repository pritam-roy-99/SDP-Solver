# NPA–BFF Toolkit: Quickstart

This notebook uses the installed `qresource` package. It demonstrates algebraic construction first, then optional CVXPY SDP runs. The numerical solver results are not rigorous dual certificates.


```python
from qresource import NPA, BFFNPA
```

## 1. CHSH Bell bound


```python
chsh = '<A0B0> + <A0B1> + <A1B0> - <A1B1>'
npa = NPA(parties=2, inputs=2, outcomes=2, level='1+AB')
print('Classical local bound:', npa.local_bound(chsh))
print('Moment words:', len(npa.word_basis()))
```

Run the following cell only after installing CVXPY and a supported solver.


```python
print(npa.maximize(chsh, solver='SCS'))
```

## 2. BFF conditional entropy from CHSH

At `bff_order=1`, the iterated mean parameter is $\alpha_1=2$. `key_settings={0: 0}` selects Alice's setting-zero outcome as the classical random variable.


```python
bff = BFFNPA(outputs=[[2,2],[2,2]], key_settings={0:0}, bff_order=1, level=2)
print('Gram words:', len(bff.basis(2)))
Q, localizing = bff.bff_polynomials()
print('Localizing polynomials:', len(localizing))
```

```python
result = bff.solve(statistics=[(chsh, '>=', 2.6)], solver='SCS', eps=1e-7)
print(result)
print('Entropy (numerical bits):', result.entropy_bits)
print('Numerically consistent:', result.reliable)
```
